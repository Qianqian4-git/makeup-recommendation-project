# 数据探索

本Notebook用于对用户上传的自拍照数据进行探索性分析，包括数据可视化和初步分析。

```python
# 导入必要的库
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# 设置绘图风格
sns.set(style='whitegrid')
```

```python
# 加载数据
data_path = '../data/processed/'
data_file = 'processed_data.csv'
df = pd.read_csv(os.path.join(data_path, data_file))

# 查看数据的前几行
df.head()
```

```python
# 数据基本信息
df.info()
```

```python
# 描述性统计
df.describe()
```

```python
# 可视化脸型分布
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='face_shape')
plt.title('脸型分布')
plt.xticks(rotation=45)
plt.show()
```

```python
# 可视化眼型分布
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='eye_shape')
plt.title('眼型分布')
plt.xticks(rotation=45)
plt.show()
```

```python
# 可视化其他特征
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x='makeup_recommendation', y='age')
plt.title('妆容推荐与年龄的关系')
plt.xticks(rotation=45)
plt.show()
```

## 结论

通过以上分析，我们可以初步了解用户的脸型、眼型及其他特征的分布情况，为后续的模型训练和妆容推荐提供基础。
